```python
# 将金融数据集的收益率计算出来作为OT
import pandas as pd
import numpy as np
import os
```

```python
data_path = '/data/nishome/user1/chaochuan/TSGym_benchmark/dataset/nasdaq/nasdaq.csv'
data = pd.read_csv(data_path)
data.columns = ['date','Close','High','Low','Open','Volume']
```

```python
data
```

```python


data['OT'] = (data['Close'].shift(-2) - data['Close']).shift(-1) / data['Close'].shift(-1)
data['OT'] = data['OT'].fillna(0)  # 填充第一个缺失值为 0
data['OT'] *= 100
```

```python
save_dirpath = '/data/nishome/user1/chaochuan/TSGym_benchmark/dataset/finance_nasdaq'
save_path = os.path.join(save_dirpath, 'finance_nasdaq.csv')
if not os.path.exists(save_dirpath):
    os.makedirs(save_dirpath)
data.to_csv(save_path, index=False)
print(f"Processed data saved to {save_path}")
```
